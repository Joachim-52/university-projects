# Welcome to learning reinforcement learning

As mentioned the practial part of the VU will concern itself with the implementation of reinforcement learning algorithms using `python` and its libraries. It is highly recomended to use a tailored `virtual enviroment` for this curse as the setup process is simplified and version missmatches can be prevented. 

Our recommendation is to use [Miniconda](https://docs.anaconda.com/miniconda/) for this purpose.
If you have not installed Miniconda or Anaconda yet, follow the installation instructions [here](https://docs.anaconda.com/miniconda/#quick-command-line-install).

As soon as you have installed Miniconda, you can create a new environment with the name `reinforcementlearning` by running
```bash
conda create -n "reinforcementlearning" python=3.11.5
```
THis will create a venv with the specified version of python but will not impact the default enviroment so no version clashes will arise.

After creating the environment, use the following command to activate it:
```bash
conda activate reinforcementlearning
```

If you open this Jupyter notebook in VSCode, select the Python interpreter in the `reinforcementlearning` conda environment in the upper right corner.
VSCode will then prompt you to install the `ipykernel` package, which you should install.

In order to install all other packages need when in the desired enviroment either execute a cell with 

```bash
!pip install packagexy
```

oder use a given requirements.txt file which we will provide. 

```bash
!pip install -r requirements
```
