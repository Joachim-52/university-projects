###  1. Write a function `calculate_revenue`

```python
# Write your code in this cell:
def calculate_revenue(crop_value:int, harvest1:int, harvest2:int=0) -> int:
    return crop_value * (harvest1 + harvest2)
```

#### 3.2 Conditions

```python
luck = [-1, -0.03, 0.01, 0.03, 1]

# Write your code here:
for l in luck:
    if l < -0.07:
        print(f"Your luck is {l}. You do not have much luck today. Today will be difficult.")
    elif -0.07 <= l < -0.02:
        print(f"Your luck is {l}. Luck is not on your side, the spirits are displeased!")
    elif -0.02 <= l < 0.02:
        print(f"Your luck is {l}. Today is going to be a neutral day.")
    elif 0.02 <= l < 0.07:
        print(f"Your luck is {l}. The luck is on your side, today you will have a bit of extra luck.")
    else:
        print(f"Your luck is {l}. The spirits are very happy today! It's going to be a great day.")
```

#### 3.3 Find the most expensive crop

```python
crop_prices = [51, 50, 250, 90]
summer_crops = ["BLUEBERRY", "CORN", "MELON", "RADISH"]

# Write your code here:
most_expensive = crop_prices[0]
summer_crop_name = summer_crops[0]

for i in range(len(crop_prices)):
    if crop_prices[i] > most_expensive:
        most_expensive = crop_prices[i]
        summer_crop_name = summer_crops[i]
        

print(summer_crop_name, most_expensive)

```

### 4. Combination

```python
# TODO Implement the task as described in 4. Combination.
crops_harvest = [
    {'name': 'Beet', 'crop_value': 200, 'harvest1': 80, 'harvest2': 40},
    {'name': 'Cranberries', 'crop_value': 150, 'harvest1': 36, 'harvest2': 136},
    {'name': 'Eggplant', 'crop_value': 120, 'harvest1': 50, 'harvest2': 30},
    {'name': 'Grape', 'crop_value': 160, 'harvest1': 60, 'harvest2': 170}
]

# Write your code here:
revenue_both = 0
for dic in crops_harvest:
    revenue_both = calculate_revenue(dic["crop_value"], dic["harvest1"], dic["harvest2"])
    dic["revenue_both"] = revenue_both

print(crops_harvest)

top_seller = crops_harvest[0]["revenue_both"]
top_seller_name = crops_harvest[0]["name"]
top_seller_crop_value = crops_harvest[0]["crop_value"]

for i in range(len(crops_harvest)):
    if top_seller < crops_harvest[i]["revenue_both"]:
        top_seller = crops_harvest[i]["revenue_both"]
        top_seller_name = crops_harvest[i]["name"]
        top_seller_crop_value = crops_harvest[i]["crop_value"]

print(f"This summer the {top_seller_name} was the crop which yielded highest revenue of {top_seller}, with the price of {top_seller_crop_value} per crop.")

total_revenue = 0
for crop in crops_harvest:
    total_revenue += crop["revenue_both"]


print(f"​​​​Total money earned this summer: {total_revenue} ")
```
