# Large Numbers in Python

```python
# 32-bit integer
i32 = 0b11111111111111111111111111111111
print(i32)
print(i32.bit_length())
```

```python
i33 = i32 + 1
print(i32)
print(i33)
print(i33.bit_length())
```

```python
print(bin(i32))
print(bin(i33))
```

```python
# 64-bit integer
i64 = 0b1111111111111111111111111111111111111111111111111111111111111111
print(i64)
print(i64.bit_length())
```

```python
i65 = i64 + 1
print(i64)
print(i65)
print(i65.bit_length())
```

```python
print(bin(i64))
print(bin(i65))
```

```python
very_large_number = 9**12345
```

```python
print(very_large_number)
```

```python
print(very_large_number.bit_length())
```

```python
print(bin(very_large_number))
```
