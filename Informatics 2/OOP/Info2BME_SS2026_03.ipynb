#### Inheritance ("Is-A")

```python
#############################
# Class Animal (Parent class)
#############################
class Animal:  
    def eat(self):
        return "Eating food"

#############################
# Class Dog (Child class)
#############################
class Dog(Animal):  # Child class inherits from Animal
    def bark(self):
        return "Woof!"

#############################
# Instantiating the Dog class
#############################
dog = Dog()
print(dog.eat())   # Output: "Eating food" (inherited from Animal)
print(dog.bark())  # Output: "Woof!" (defined in Dog)
```

```python
###############################
# Class Vehicle (Parent class)
###############################
class Vehicle:
    def move(self):
        return "The vehicle moves"

###############################
# Class Car (Child class)
###############################
class Car(Vehicle):
    def drive(self):
        return super().move() + " smoothly on the road."

###############################
# Instantiating the Car class
###############################
my_car = Car()
print(my_car.drive())  # Output: "The vehicle moves smoothly on the road." ("The vehicle moves" inherited from Vehicle using super(); super() is used to call a parent class method)
```

#### Composition ("Has-A")

```python
####################
# Class Engine
####################
class Engine:
    '''
    Engine class represents the engine of a car.
    '''

    def __init__(self, year):
        '''
        Initialize an engine object.
        '''
        self.year = year

    def start(self):
        '''
        Start the engine.
        '''
        return "Engine started"

####################
# Class Car
####################
class Car:
    ''' 
    Car class represents a car.
    '''

    def __init__(self, year):
        '''
        Initialize a car object.
        '''
        self.engine = Engine(year=year)  # Car HAS an Engine (Composition)

    def drive(self):
        '''
        Drive the car.
        '''
        return self.engine.start() + " → Car is moving!"  # Delegation inside composition

###########################
# Instantiate a Car object
###########################
my_car = Car(year=2025)
my_car.drive()  # Output: "Engine started → Car is moving!"
```

#### Duck-Typing ("Acts like")

```python
####################
# Class Bird
####################
class Bird:
    def fly(self):
        return "Flapping wings"

####################
# Class Airplane
####################
class Airplane:
    def fly(self):
        return "Jet engines roaring"

########################
# Class FlightSimulator
########################
class FlightSimulator:
    def __init__(self, flying_object):
        self.flying_object = flying_object  # Delegation (storing any object (that "flies") )

    def simulate(self):
        return self.flying_object.fly()  # Delegation in action (calling the fly method of the object)

# Instantiate Bird and Airplane objects
bird = Bird()
plane = Airplane()

# Instantiate FlightSimulator objects
sim1 = FlightSimulator(bird)
sim2 = FlightSimulator(plane)

# Simulate flying
print(sim1.simulate())  # Output: "Flapping wings"
print(sim2.simulate())  # Output: "Jet engines roaring"
```

#### Same Class Student using best coding practices

```python
### Class Student
class StudentEnhanced:
   '''
    Class StudentEnhanced represents a student.
    It is written in an enhanced way (with docstrings).
   '''
   ### Constructor
   def __init__(self, id: int, firstname: str, lastname: str, email: str):
      '''   
        Constructor of the class Student

        Parameters:
        id (int): The matriculation number of the student
        firstname (str): The first name of the student
        lastname (str): The last name of the student
        email (str): The email address of the student
      '''
      self.id = id
      self.firstname = firstname       # Public attribute (by default, all attributes are public)
      self.lastname = lastname         # Public attribute (by default, all attributes are public)
      self._email = email              # Private attribute (by convention, it is prefixed with an underscore)

   ### Return all the main info about the student
   def info(self):
      '''
        Return all the main info about the student

        Returns:
        str: A string containing the matriculation number, first name, last name and email address of the student
      '''
      return f"Matriculation #: {self.id} ({self.firstname} {self.lastname}, {self._email})"
   
   ### Set the first name of the student
   def set_firstname(self, firstname):
      '''
         Set the first name of the student
   
         Parameters:
         firstname (str): The first name of the student
      '''
      self.firstname = firstname

   ### Get the first name of the student
   def get_firstname(self):
      '''
         Get the first name of the student
   
         Returns:
         str: The first name of the student
      '''
      return self.firstname
   
   ### Get the email address of the student as a property (the pythonic way)
   @property
   def email(self):
      '''
         Get the email address of the student
   
         Returns:
         str: The email address of the student
      '''
      return self._email

   ### Set the email address of the student as a property (the pythonic way)
   @email.setter
   def email(self, email):
      '''
         Set the email address of the student
   
         Parameters:
         email (str): The email address of the student
      '''
      # You can add validation or other custom logic here
      self._email = email
```

```python
# Instanciating an object of type (class) StudentEnhanced
s2 = StudentEnhanced(id = 1234, 
                     firstname = "Thomas", 
                     lastname = "Müller", 
                     email = "thomas.mueller@tugraz.at")
s2.info()
```

## Enrollment system

```python
### Class User
class User(object):
    def __init__(self, username, email):
        self.username = username
        self.email = email

    def info(self):
        return f"User {self.username} ({self.email})"
    
### Class Student (inherits from User)
class Student(User):
    def __init__(self, id, name, user, email, year):
        super().__init__(user, email)
        self.id = id
        self.name = name
        self.year = year

    def info(self):
        return f"Matriculation ID {self.id} ({self.year}): {self.name}, {self.username}, {self.email}"

### Class Course
class Course(object):
    def __init__(self, code, name):
        self.code = code
        self.name = name
    
    def info(self):
        return f"({self.code}) {self.name}"

### Class Enrollment (composition of Course and Student)
class Enrollment(object):
    def __init__(self, course, term, year):
        self.course = course
        self.term = term
        self.year = year
        self.students = []

    def enroll(self, student):
        self.students.append(student)

    def info(self):
        print(f"{len(self.students)} students are enrolled in {self.course.name} ({self.course.code}) in {self.term} {self.year}")
    
    def extended_info(self):
        print(f"{len(self.students)} enrolled students:")
        for student in self.students:
            print(student.info())
    
```

```python
# Students
s1 = Student(1, "Thomas Müller", "tm822", "thomas.mueller@tugraz.at", 2024)
s2 = Student(2, "Manuel Neuer", "mn222", "manuel.neuer@tugraz.at", 2024)
s3 = Student(3, "Angela Merkel", "am091", "angela.merkel@tugraz.at", 2024)
```

```python
# Courses
c1 = Course("CS123", "Python Programming")
c2 = Course("CS456", "Web Development")
c3 = Course("CS789", "Data Science")
```

```python
# Enrollment in Course #1
e1 = Enrollment(c1, "WS", 2025)
e1.enroll(s1)
e1.enroll(s2)
e1.info()
e1.extended_info()
```

```python
# Enrollment in Course #2
e2 = Enrollment(c2, "SS", 2025)
e2.enroll(s1)
e2.enroll(s2)
e2.enroll(s3)
e2.info()
e2.extended_info()
```

### Instance vs. Reference

```python
## Example of an instance:

x = 10          # x is an instance of an integer
print(type(x))  # Output: <class 'int'>
```

```python
### Example of an immutable object reference and the use of the id() function:

x = 10                  # x is an instance of an integer
y = x                   # y is a reference to the same integer object as x

### Using id: it returns the memory address of the object
print(id(x))            # print the memory address of x
print(id(y))            # print the memory address of y

### Using hex and id together: it returns the memory address of the object in hexadecimal format
print(hex(id(x)))       # print the memory address of x in hexadecimal format
print(hex(id(y)))       # print the memory address of y in hexadecimal format

### Using the assert statement to verify whether x and y reference the same object
assert x is y                    # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)
assert id(x) == id(y)            # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)
assert hex(id(x)) == hex(id(y))  # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)
```

```python
y += 1                  # y is now a reference to a new integer object (x+1)
print(x)                # Output: 10
print(y)                # Output: 11

assert x is y                    # This assertion will fail, as x and y reference different objects
assert id(x) == id(y)            # This assertion will fail, as x and y reference different objects
assert hex(id(x)) == hex(id(y))  # This assertion will fail, as x and y reference different objects
```

```python
### Alternative way to verify whether x and y reference the same object

if x is y:
    print("[is] x and y reference the same object")
else: 
    print("[is] x and y reference different objects")

if id(x) == id(y):
    print("[id] x and y reference the same object")
else:
    print("[id] x and y reference different objects")

if hex(id(x)) == hex(id(y)):
    print("[hex-id] x and y reference the same object")
else:
    print("[hex-id] x and y reference different objects")
```

```python
### Example of a mutable object (list) and the use of the id() function:
x = [1, 2, 3]           # x is an instance of a list
y = x                   # y is a reference to the same list object as x

### Using id: it returns the memory address of the object
print("-- y points to x --")
print(x, id(x), hex(id(x)))            # print the memory address of x
print(y, id(y), hex(id(y)))            # print the memory address of y

### Using the assert statement to verify whether x and y reference the same object
assert x is y                    # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)
assert id(x) == id(y)            # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)
assert hex(id(x)) == hex(id(y))  # verify whether x and y reference the same object (if True, the assertion passes. If False, an AssertionError is raised)

### Modifying the list object
print("-- y adds one element --")
y.append(4)             # y is now a reference to the same list object as x, with an additional element (4)
print(f'x: {x} {id(x)} {hex(id(x))}')        # Output: [1, 2, 3, 4]
print(f'y: {y} {id(y)} {hex(id(y))}')        # Output: [1, 2, 3, 4]

assert x is y                    # This assertion will pass, as x and y reference the same object
assert id(x) == id(y)            # This assertion will pass, as x and y reference the same object
assert hex(id(x)) == hex(id(y))  # This assertion will pass, as x and y reference
```

### Comparing variables

```python
x = 100  # x is an integer
y = 200  # y is an integer
z = x    # z points to the same object as x

# By value:
print("Comparison by value:")
print(x == y)   # returns False
print(x == z)   # returns True

# By reference:
print("Comparison by identity:")
print("x is y: ", x is y)                    # returns False
print('id(x) == id(y): ', id(x) == id(y))    # returns False

print(f"x is z: {x is z}")                   # returns True (print formatting using f-string)
print(f"x is not y: {x is not y}")           # returns True (print formatting using f-string)
print(f"id(x) == id(z): {id(x) == id(z)}")   # returns True
```

```python
x = [4, 5, 6]  # x is a list 
y = [4, 5, 6]  # y is a list
z = x          # z now refers to the same object as x

print(x == y)  # True  (Same values)
print(x is y)  # False (Different objects)
print(x is z)  # True  (Same object in memory)
```

### Namespaces

```python
var_1 = 10  # Global namespace

def my_function():
    var_2 = 5  # Local namespace (exists only inside the function)
    print(var_2)

print(var_1)  # Output: 10

my_function()  # Output: 5

try:
    print(var_2)  # Error! var_2 is not defined globally, only locally
except Exception as e:
    print("[ERROR]", e)
```

```python
var_3 = 10  # Global namespace

def my_function():
    var_3 = 5        # Local namespace (exists only inside the function)
    print(var_3)     # Output: 5 (because var_3 is defined locally, it is not the same as the global var_3)

my_function()        # Output: 5
print(var_3)         # Output: 10
```

```python
var_4 = 10  # Global namespace

def my_function():
    global var_4    # Declare var_4 as global
    var_4 = 5       # Global var_4 is now updated
    print(var_4)

print(var_4)  # Output: 10
my_function()
print(var_4)  # Output: 5
```

```python
## NON-local vs. local variables

def outer_function():
    x = "outer"

    def inner_function_1():
        x = "inner 1"

        def inner_function_2():
            nonlocal x
            x = "inner 2"

        inner_function_2()
        print("Value of x in inner_function_1:", x)

    inner_function_1()
    print("Value of x in outer_function:", x)

outer_function()
```
