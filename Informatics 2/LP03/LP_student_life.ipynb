````python
# Your code goes here:
class Activity:
    def __init__(self, name:str, hours:int, required_items:set[str]):
        self.name = name
        self.hours = hours
        self.required_items = required_items

    def __str__(self) -> str:
        return f"{self.name} for ({self.hours} hours)"

class ActivityExecutionError(Exception):
    pass

class Student:
    def __init__(self, number:str, backpack_content:set[str]):
        self.number = number
        self.backpack_content = backpack_content
        self.activity_log = []


    @property
    def number(self) -> str:
        return self._number
    
    @number.setter
    def number(self, number:str):
        if not number.isnumeric():
            raise NameError("Student number is not numeric!")
        else:
            self._number = number
    
    @property
    def spare_time(self):
        total = 0
        for activity in self.activity_log:
            total += activity.hours
        return 24 - total
    
    def all_items_packed(self, activity:Activity):
        return activity.required_items.issubset(self.backpack_content)
    
    def enough_time_left(self, activity:Activity):
        if self.spare_time >= activity.hours:
            return True
        else:
            return False
        
    def execute(self, activity:Activity):
        if not self.enough_time_left(activity):
            raise ActivityExecutionError("The day is over!")
        if not self.all_items_packed(activity):
            raise ActivityExecutionError(f"Not all items available for {activity.name}!")
        self.activity_log.append(activity)

    def live_day(self, todo:list[Activity]):
        for to in todo:
            try:
                self.execute(to)
                print(f"{self.number} executed {to}. Spare time left: {self.spare_time} hours.")
            except ActivityExecutionError as error:
                print(f"{self.number} can not execute {to}!")
                print(error)
                return
        
````
